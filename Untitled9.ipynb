```python
import cv2
import matplotlib.pyplot as plt

# Step 1: Load the image
img = cv2.imread("iron.jpg")

if img is None:
    print("Error: Image not found. Check the file path.")
else:
    # Step 2: Convert BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Step 3: Convert into Grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 4: Global Thresholding
    _, global_thresh = cv2.threshold(
        gray, 127, 255, cv2.THRESH_BINARY
    )

    # Step 5: Adaptive Mean Thresholding
    adaptive_mean = cv2.adaptiveThreshold(
        gray, 255,
        cv2.ADAPTIVE_THRESH_MEAN_C,
        cv2.THRESH_BINARY,
        11, 2
    )

    # Step 6: Adaptive Gaussian Thresholding
    adaptive_gaussian = cv2.adaptiveThreshold(
        gray, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        11, 2
    )

    # Step 7: Otsu's Thresholding
    otsu_value, otsu_thresh = cv2.threshold(
        gray, 0, 255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # Step 8: Display all outputs
    titles = [
        "Original Image",
        "Grayscale Image",
        "Global Thresholding",
        "Adaptive Mean",
        "Adaptive Gaussian",
        "Otsu's Thresholding"
    ]

    images = [
        img_rgb,
        gray,
        global_thresh,
        adaptive_mean,
        adaptive_gaussian,
        otsu_thresh
    ]

    plt.figure(figsize=(12, 8))

    for i in range(6):
        plt.subplot(2, 3, i + 1)

        if i == 0:
            plt.imshow(images[i])
        else:
            plt.imshow(images[i], cmap="gray")

        plt.title(titles[i])
        plt.axis("off")

    plt.tight_layout()
    plt.show()

    print("Otsu Threshold Value:", otsu_value)
    print("Image Segmentation Completed Successfully!")
```
